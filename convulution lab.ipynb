```python
import numpy as np

# Input matrix (5x5)
matrix1 = np.array([
    [3, 1, 0, 2, 1],
    [2, 0, 1, 3, 0],
    [4, 1, 0, 1, 2],
    [1, 3, 2, 0, 1],
    [0, 2, 1, 4, 2]
])

# 3x3 kernel
kernel = np.array([
    [1, 0, -1],
    [1, 0, -1],
    [1, 0, -1]
])

# Create a 3x3 output matrix
output = np.zeros((3, 3), dtype=int)

# Apply the 3x3 sliding window
for i in range(3):
    for j in range(3):

        # Extract the 3x3 window
        window = matrix1[i:i+3, j:j+3]

        # Multiply corresponding elements and sum
        output[i, j] = np.sum(window * kernel)

        # Display the current window and result
        print(f"\nWindow at position ({i}, {j}):")
        print(window)
        print("Result:", output[i, j])

# Display final output matrix
print("\nFinal Output Matrix:")
print(output)
```

```python
import numpy as np

cat = np.array([0.3, 0.1, 0.2, 0.1])
dog = np.array([0.1, 0.2, 0.1, 0.3])
bird = np.array([0.0, 0.4, 0.1, 0.2])

# Given matrix
matrix = np.array([
    [8, 0],
    [4, 1]
])

# Reshape arrays into 2x2 matrices
cat_matrix = cat.reshape(2, 2)
dog_matrix = dog.reshape(2, 2)
bird_matrix = bird.reshape(2, 2)

# Multiply and sum all values
cat_result = np.sum(cat_matrix * matrix)
dog_result = np.sum(dog_matrix * matrix)
bird_result = np.sum(bird_matrix * matrix)

# Display results
print("Cat result:", cat_result)
print("Dog result:", dog_result)
print("Bird result:", bird_result)

# Find the predicted class
results = {
    "Cat": cat_result,
    "Dog": dog_result,
    "Bird": bird_result
}

print("\nPredicted class:", max(results, key=results.get))
```
